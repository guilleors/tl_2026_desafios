## Descuento precio

Escribir una función que aplique un descuento a un precio y otra que aplique el IVA a un precio. Escribir una tercera función que reciba un diccionario con los precios y porcentajes de una cesta de la compra, y una de las funciones anteriores, y utilice la función pasada para aplicar los descuentos o el IVA a los productos de la cesta y devolver el precio final de la cesta

```python
def aplicar_descuento(precio, porcentaje):
    descuento = precio * (porcentaje / 100)
    precio_final = precio - descuento
    return precio_final

def aplicar_iva(precio, porcentaje):
    iva = precio * (porcentaje / 100)
    precio_final = precio + iva
    return precio_final

def procesar_cesta(cesta, funcion_a_usar):
    total = 0
    for precio in cesta:
        porcentaje = cesta[precio]
        precio_calculado = funcion_a_usar(precio, porcentaje)
        total = total + precio_calculado
    return total


cesta = {100: 10, 50: 20}
print("Total con descuento:", procesar_cesta(cesta, aplicar_descuento))
```

## Calculadora

Escribir una función que simule una calculadora científica que permita calcular el seno, coseno, tangente, exponencial y logaritmo neperiano. La función preguntará al usuario el valor y la función a aplicar, y mostrará por pantalla una tabla con los enteros de 1 al valor introducido y el resultado de aplicar la función a esos enteros

```python
import math

def calculadora():
    print("Opciones: seno, coseno, tangente, exponencial, logaritmo")
    funcion = input("¿Qué función quieres usar?: ")
    numero = int(input("Introduce un número entero: "))

    print("Número | Resultado")
    print("------------------")

    for i in range(1, numero + 1):
        if funcion == "seno":
            resultado = math.sin(i)
        elif funcion == "coseno":
            resultado = math.cos(i)
        elif funcion == "tangente":
            resultado = math.tan(i)
        elif funcion == "exponencial":
            resultado = math.exp(i)
        elif funcion == "logaritmo":
            resultado = math.log(i)
        
        print(i, "|", resultado)

calculadora()
```

## Aplicar función

Escribir una función que reciba otra función y una lista, y devuelva otra lista con el resultado de aplicar la función dada a cada uno de los elementos de la lista

```python
def aplicar_funcion(funcion, lista):
    nueva_lista = []
    for elemento in lista:
        resultado = funcion(elemento)
        nueva_lista.append(resultado)
    return nueva_lista


def duplicar(numero):
    return numero * 2

mis_numeros = [1, 2, 3, 4]
print(aplicar_funcion(duplicar, mis_numeros))
```

## Filtro true

Escribir una función que reciba otra función booleana y una lista, y devuelva otra lista con los elementos de la lista que devuelvan True al aplicarles la función booleana

```python
def filtrar_lista(funcion_booleana, lista):
    nueva_lista = []
    for elemento in lista:
        # Si la función devuelve True, guardamos el elemento
        if funcion_booleana(elemento) == True:
            nueva_lista.append(elemento)
    return nueva_lista


def es_par(numero):
    return numero % 2 == 0

mis_numeros = [1, 2, 3, 4, 5, 6]
print(filtrar_lista(es_par, mis_numeros))
```

## Diccionario palabras longitud

Escribir una función que reciba una frase y devuelva un diccionario con las palabras que contiene y su longitud

```python
def medir_palabras(frase):
    palabras = frase.split()  # Separa la frase en una lista de palabras
    diccionario = {}
    
    for palabra in palabras:
        longitud = len(palabra)
        diccionario[palabra] = longitud  # Guarda 'palabra': longitud
        
    return diccionario


resultado = medir_palabras("Hola mundo desde Python")
print(resultado)
```

## Lista calificaciones

Escribir una función reciba una lista de notas y devuelva la lista de calificaciones correspondientes a esas notas

```python
def evaluar_notas(lista_de_notas):
    calificaciones = []
    
    for nota in lista_de_notas:
        if nota < 5:
            calificaciones.append("Suspenso")
        elif nota < 7:
            calificaciones.append("Aprobado")
        elif nota < 9:
            calificaciones.append("Notable")
        else:
            calificaciones.append("Sobresaliente")
            
    return calificaciones


mis_notas = [4, 6, 8.5, 10]
print(evaluar_notas(mis_notas))
```

## Diccionario calificaciones

Escribir una función reciba un diccionario con las asignaturas y las notas de un alumno y devuelva otro diccionario con las asignaturas en mayúsculas y las calificaciones correspondientes a las notas

```python
def calificar_asignaturas(diccionario_notas):
    resultado = {}
    
    for asignatura in diccionario_notas:
        nota = diccionario_notas[asignatura]
        
        # Convertir nombre a mayúsculas
        asignatura_mayuscula = asignatura.upper()
        
        # Determinar la calificación
        if nota < 5:
            calificacion = "Suspenso"
        elif nota < 7:
            calificacion = "Aprobado"
        elif nota < 9:
            calificacion = "Notable"
        else:
            calificacion = "Sobresaliente"
            
        resultado[asignatura_mayuscula] = calificacion
        
    return resultado


notas_alumno = {'Matemáticas': 8, 'Historia': 4}
print(calificar_asignaturas(notas_alumno))
```

## Módulo vector

Escribir una función que calcule el módulo de un vector

```python
import math

def modulo_vector(vector):
    suma_cuadrados = 0
    
    for numero in vector:
        suma_cuadrados = suma_cuadrados + (numero ** 2)
        
    resultado = math.sqrt(suma_cuadrados)
    return resultado


mi_vector = [3, 4]
print("Módulo:", modulo_vector(mi_vector))
```

## Valores atípicos

Escribir una función que reciba una muestra de números y devuelva los valores atípicos, es decir, los valores cuya puntuación típica sea mayor que 3 o menor que -3. Nota: La puntuación típica de un valor se obtiene restando la media y dividiendo por la desviación típica de la muestra

```python
import math

def obtener_valores_atipicos(muestra):
    suma_total = 0
    for x in muestra:
        suma_total = suma_total + x
    media = suma_total / len(muestra)
    
    suma_diferencias = 0
    for x in muestra:
        suma_diferencias = suma_diferencias + ((x - media) ** 2)
    desviacion = math.sqrt(suma_diferencias / (len(muestra) - 1))
    
    atipicos = []
    for x in muestra:
        puntuacion_tipica = (x - media) / desviacion
        
        if puntuacion_tipica > 3 or puntuacion_tipica < -3:
            atipicos.append(x)
            
    return atipicos


datos = [10, 12, 11, 12, 11, 14, 100, 12, -80]
print("Valores atípicos:", obtener_valores_atipicos(datos))
```
