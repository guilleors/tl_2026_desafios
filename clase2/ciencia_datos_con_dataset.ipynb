EDA (Análisis Exploratorio de Datos): distribuciones, outliers y relaciones entre variables

```python
# Distribución de year
plt.figure(figsize=(10,6))
sns.histplot(df['year'], bins=15, kde=True)
plt.title("Distribución de year")
plt.xlabel("Años (%)")
plt.ylabel("Frecuencia")
plt.show()
```

```python
# Boxplot de language
plt.figure(figsize=(10,6))
sns.boxplot(x=df['language'])
plt.title("Boxplot de language")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.barplot(
    data=df.head(10),
    x="num_pushers",
    y="language_type",
    hue="num_pushers",       # asignamos hue
    palette="viridis",
    legend=False          # evita mostrar leyenda duplicada
)
plt.title("Top 10 Años por Cantidad de Aportantes de Código")
plt.xticks(rotation=45)
plt.show()
```

Otras Visualizaciones

```python
# Violin plot por language_type (ajustado a nuevas versiones de Seaborn)
plt.figure(figsize=(12,6))
sns.violinplot(
    data=df,
    x="language_type",
    y="year",
    hue="language_type",        # asignamos hue para usar palette
    palette="muted",
    legend=False           # evita mostrar leyenda duplicada
)
plt.title("Distribución de Tipo de Lenguaje")
plt.xticks(rotation=45)
plt.show()
```
